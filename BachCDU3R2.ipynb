# Conclusión

En esta práctica utilizamos Google Colab y el lenguaje R para analizar los datos del archivo Concentrado.xlsx, el cual contiene las respuestas de un test de personalidad aplicado a un grupo de 489 alumnos, quienes tienen en promedio 24.5 años de edad. Al ejecutar los comandos summary() y describe(), logré obtener las estadísticas del grupo.
Esto fue lo más importante que encontré en los resultados:
 Lo más alto: El rasgo en el que el grupo obtuvo puntajes más altos fue Responsabilidad (Tesón) con un promedio de 33.52, seguido por Cordialidad / Amabilidad con 32.15. Esto significa que, en general, los alumnos se consideran responsables, organizados y empáticos.
 El punto medio: Los rasgos de Apertura a la Experiencia (promedio de 41.38) y Extraversión (promedio de 29.92) se mantuvieron en niveles intermedios y equilibrados.
 Lo más bajo: El promedio más bajo de todo el grupo fue el de Estabilidad Emocional, con apenas 20.02. Esto nos dice que una gran parte de los alumnos suele lidiar con niveles altos de estrés, sensibilidad o ansiedad.
